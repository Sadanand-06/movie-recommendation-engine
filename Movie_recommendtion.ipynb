```python
# Load the movies and credits data
movies = pd.read_csv('/content/tmdb_5000_movies.csv')
credits = pd.read_csv('/content/tmdb_5000_credits.csv')  # Assuming 'tmdb_5000_credits.csv' contains cast info

# Merge 'cast' and 'crew' columns from 'credits' DataFrame into 'movies' DataFrame
movies = pd.merge(movies, credits[['movie_id','title', 'cast', 'crew']], on='title', how='left') # Include 'movie_id' here

# Now you can select the desired columns, including 'movie_id', 'cast', and 'crew'
#... (rest of the data processing code)

# Ensure 'genres' is a list of strings
movies['genres'] = movies['genres'].apply(convert)

# Convert all list columns to strings of space-separated values
for col in ['overview', 'genres', 'keywords', 'cast', 'crew']:
    movies[col] = movies[col].apply(lambda x: ' '.join(map(str, x)) if isinstance(x, list) else str(x))

# Concatenate columns into 'tags' column
movies['tags'] = movies['overview'] + ' ' + movies['genres'] + ' ' + movies['keywords'] + ' ' + movies['cast'] + ' ' + movies['crew']

# Create the new DataFrame, including 'movie_id'
new_df = movies[['movie_id', 'title', 'tags']] # Now includes 'movie_id'
```

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score

def evaluate_recommendations(user_ratings, recommendations):
    """
    Evaluates the recommendations using precision and recall.

    Args:
        user_ratings: A dictionary where keys are user IDs and values are lists of movie IDs the user has rated positively.
        recommendations: A dictionary where keys are user IDs and values are lists of movie IDs recommended to the user.

    Returns:
        A tuple containing the average precision and recall across all users.
    """

    precisions = []
    recalls = []

    for user_id in user_ratings:
        if user_id in recommendations:
            relevant_items = set(user_ratings[user_id])
            recommended_items = set(recommendations[user_id])

            # Calculate precision and recall for the current user
            precision = precision_score(
                [1 if item in relevant_items else 0 for item in recommended_items],
                [1] * len(recommended_items),  # Assume all recommendations are positive
                zero_division=0  # Handle cases where no recommendations are made
            )
            recall = recall_score(
                [1 if item in relevant_items else 0 for item in relevant_items],
                [1 if item in recommended_items else 0 for item in relevant_items],
                zero_division=0  # Handle cases where no relevant items exist
            )

            precisions.append(precision)
            recalls.append(recall)

    # Calculate average precision and recall across all users
    avg_precision = np.mean(precisions) if precisions else 0
    avg_recall = np.mean(recalls) if recalls else 0

    return avg_precision, avg_recall

# Example Usage:
user_ratings = {
    1: [1, 2, 3],  # User 1 has rated movies 1, 2, and 3 positively
    2: [4, 5],  # User 2 has rated movies 4 and 5 positively
}
recommendations = {
    1: [2, 3, 6],  # Movies recommended to user 1
    2: [4, 7],  # Movies recommended to user 2
}

precision, recall = evaluate_recommendations(user_ratings, recommendations)
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
```
